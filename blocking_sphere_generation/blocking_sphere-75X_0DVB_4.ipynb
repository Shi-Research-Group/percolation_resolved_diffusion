```python
spacing = 0.5
mask_thickness=0.0
h= 0.1
threshold_abs= 0.3
preserve_threshold=103.0
smooth_threshold=106.0
max_smooth_radius=2.0
extra_padding= 2
return_dgrid = True
min_distance= 1.0
min_region_size=0.00005
minimum_window_separation=1.5
```

```python
mat_atoms = read('./75X_0DVB_4.cif',store_tags=True)
```

```python
site_labels = mat_atoms.info['_atom_site_label']
radii = trappe_dia[site_labels].values/ 2.0
dgrid = mgr.dgrid_from_atoms_rust(mat_atoms, spacing=spacing, radii=radii)
```

```python
results = mgr.do_periodic_segmentation_rust(
                    dgrid, 
                    mat_atoms,
                    mask_thickness=mask_thickness,
                    h=h,
                    spacing=spacing,
                    return_dgrid=False,
                    threshold_abs=threshold_abs,
                    min_distance=min_distance,
                    return_padded_regions=True,
                    preserve_threshold=preserve_threshold,
                    smooth_threshold=smooth_threshold,
                    max_smooth_radius=max_smooth_radius,
                    extra_padding=extra_padding,
                    compute_connections=True,
                    min_region_size=min_region_size,
                    minimum_window_separation=minimum_window_separation,
                    watershed_backend="rust_det"
                )
```

```python
regions = results["regions"]
all_maxima = results["all_maxima"]
maxima_info = results["maxima_info"]
connections = results["connections"]
dgrid_1d = dgrid.flatten()
regions_1d = regions.flatten()
maxima = all_maxima[: regions.max()]
centroid_coordinate=mgr.find_coord_from_indices(maxima,regions.shape,mat_atoms)
all_maxima_radii = dgrid[tuple(maxima_info[:, 1:4].T)]
maxima_radii = all_maxima_radii[: regions.max()]
```

```python
graph = mgr.make_directed_graph_from_segmentation_results(results)
base_labels = mgr.build_base_region_labels_from_results(results)
plr = mgr.find_pore_limiting_radius_winding(
    results, graph=graph, base_region_labels=base_labels,include_dangling=True
)

if plr is None:
    plr=0

pld=plr*2

if pld >= 3.73:
    axis_clusters, cluster_details, cluster_metadata = mgr.percolation_all_axes_winding(
    results, threshold=1.865, graph=graph, base_region_labels=base_labels,include_dangling=True
    )
```

```python
percolation_regions = np.array(
    axis_clusters['x'] + axis_clusters['y'] + axis_clusters['z']
).flatten()
percolation_regions = np.unique(percolation_regions, axis=0)

grid, shape = mgr.dask_grid_over_atoms(mat_atoms, spacing=spacing)
grid_coordinate=grid.compute()

mask = np.zeros(len(regions_1d), dtype=bool)
for i, n in enumerate(regions_1d):
    if n in percolation_regions and dgrid_1d[i] >= 1.865:
        mask[i] = True

grid_coordinate_percolation = grid_coordinate[mask]
grid_distance_percolation=dgrid_1d[mask]
N = len(grid_coordinate_percolation)
```

```python
output_path = "./75X_0DVB_4.block"
with open(output_path, "w") as f:
    f.write(f"{N}\n\n")
    for point in grid_coordinate_percolation:
        f.write(f"N    {point[0]:.12f}    {point[1]:.12f}    {point[2]:.12f}\n")
```

```python
#find the initial spheres
initial_blocking_sphere_center=centroid_coordinate[percolation_regions-1]
initial_blocking_sphere_center_scaled=mgr.get_fractional_coordinates(initial_blocking_sphere_center, mat_atoms)
initial_blocking_sphere_radii=maxima_radii[percolation_regions-1]

#find the points in the percolating region that fall beyond the initial spheres
centroids = initial_blocking_sphere_center 
radii = initial_blocking_sphere_radii.flatten()
mask = np.ones(len(grid_coordinate_percolation), dtype=bool)
for c, r in zip(centroids, radii):
    dist2 = np.sum((grid_coordinate_percolation - c)**2, axis=1)  
    mask &= dist2 > r**2
grid_coordinate_percolation_outside = grid_coordinate_percolation[mask]
grid_distance_percolation_outside=grid_distance_percolation[mask]


#find other spheres
r = [] 
c=[]
while len(grid_coordinate_percolation_outside) > 0:
    idx = np.argmax(grid_distance_percolation_outside)
    blocking_sphere_center = grid_coordinate_percolation_outside[idx]
    blocking_sphere_center_scaled=mgr.get_fractional_coordinates(blocking_sphere_center, mat_atoms)
    blocking_sphere_radii = grid_distance_percolation_outside[idx]

    c.append(blocking_sphere_center_scaled)
    r.append(blocking_sphere_radii)


    mask = np.ones(len(grid_coordinate_percolation_outside), dtype=bool)
    dist2 = np.sum(( grid_coordinate_percolation_outside - blocking_sphere_center)**2, axis=1)
    mask &= dist2 > blocking_sphere_radii**2
    grid_coordinate_percolation_outside = grid_coordinate_percolation_outside[mask]
    grid_distance_percolation_outside = grid_distance_percolation_outside[mask]
```

```python
c=np.array(c)
r=np.array(r)

file_path = f"./75X_0DVB_4.block"
with open(file_path, 'w') as blocking_file:
    blocking_file.write(f"{len(initial_blocking_sphere_center_scaled)+len(c)}\n")
    for i, s in enumerate(initial_blocking_sphere_center_scaled):
        blocking_file.write(f"{s[0]}\t{s[1]}\t{s[2]}\t{initial_blocking_sphere_radii[i]}\n")
    for i, s in enumerate(c):
        blocking_file.write(f"{s[0]}\t{s[1]}\t{s[2]}\t{r[i]}\n")  
```
